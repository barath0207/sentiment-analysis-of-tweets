## Objective
Classify tweets into Positive, Negative, and Neutral sentiment using NLP preprocessing, TF-IDF, and Logistic Regression.

```python
from pathlib import Path
import pandas as pd
import sys
sys.path.append(str(Path.cwd().parent / 'src'))
from sentiment_analysis import clean_text
```

```python
data_path = Path.cwd().parent / 'data' / 'tweets.csv'
df = pd.read_csv(data_path)
df.head()
```

```python
df['clean_tweet'] = df['tweet'].apply(clean_text)
df[['tweet', 'clean_tweet', 'sentiment']].head(10)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

X_train, X_test, y_train, y_test = train_test_split(
    df['clean_tweet'], df['sentiment'], test_size=0.20,
    random_state=42, stratify=df['sentiment']
)

model = Pipeline([
    ('tfidf', TfidfVectorizer(ngram_range=(1, 2))),
    ('classifier', LogisticRegression(max_iter=1000, random_state=42))
])
model.fit(X_train, y_train)
pred = model.predict(X_test)
print('Accuracy:', accuracy_score(y_test, pred))
print(classification_report(y_test, pred))
```

```python
examples = [
    'I love this new update!',
    'This service is terrible and frustrating.',
    'The meeting starts at 5 PM today.'
]
for tweet in examples:
    print(tweet, '->', model.predict([clean_text(tweet)])[0])
```
